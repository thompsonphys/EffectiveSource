```python
def ELX(M, a, p, e):

    if a > 0:
        signA = 1.0
    else:
        signA = -1.0

    p2 = p * p
    p3 = p2 * p
    e2p3 = 3.0 + e * e

    C = a * a - p
    C *= C
    N = 2.0 / p * (-p2 + (e2p3 - a * a) * p - a * a * (1 + 3.0 * e * e))
    F = (
        1.0
        / p3
        * (
            p3
            - 2.0 * e2p3 * p2
            + e2p3 * e2p3 * p
            - 4.0 * a * a * (1.0 - e * e) * (1.0 - e * e)
        )
    )

    X = np.sqrt((-N - signA * np.sqrt(N * N - 4.0 * F * C)) / (2.0 * F))
    E = np.sqrt(1 - (M / p) * (1.0 - e * e) * (1 - X * X / (p * p) * (1.0 - e * e)))
    L = X + a * E

    return E, L, X


def fourvel(a, p, e, psi, X, E):
    X2 = X * X
    return (
        e
        * np.sin(psi)
        / p
        * (X2 + a * a + 2.0 * X * a * E - 2.0 * X2 / p * (3.0 + e * np.cos(psi)))
    )
```

```python
M = 1.0
a = 0.5
p = 10.0
e = 0.1
x = 1

rMin = p / (1.0 + e)
rMax = p / (1.0 - e)

E, L, X = ELX(M, a, p, e)
ur = fourvel(a, p, e, 0.0, X, E)
```

```python
coord_p = ef.make_coordinate(t=0.0, r=p, theta=np.pi/2.0, phi = 0.0)
ef.effsource_init(M, a)
ef.effsource_set_particle(coord_p, E, L, ur)
```
